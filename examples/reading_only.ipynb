# Reading only

Read documents into passages that carry where they came from, with no Neo4j. The documents are
invented samples that `graphme.samples` writes. Needs `pip install "graph-me-maybe[langchain]"`
for the last cell, and a provider key.

```python
import os
from pathlib import Path

import graphme
from graphme.samples import write_samples

# Your provider and its key. Pages and text are sent to it; nothing else leaves this machine.
SETTINGS = {"provider": "openai", "api_key": os.environ["OPENAI_API_KEY"]}
```

```python
folder = Path("samples")
files = write_samples(folder)
[f.name for f in files]
```

```python
doc = graphme.read(folder / "Leak Test Record.xlsx", **SETTINGS)
for passage in graphme.passages(doc):
    print(passage.metadata, "
", passage.text[:200], "
")
```

The same passages as LangChain documents, for any vector store:

```python
from graphme.langchain import GraphMeLoader

docs = GraphMeLoader(folder / "GS-110_Leak Test Specification.docx", **SETTINGS).load()
docs[0]
```
