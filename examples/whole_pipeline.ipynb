# The whole pipeline

Put a folder in, ask questions, get answers that cite files, sheets, cells and pages. The library
runs its own Neo4j in its home folder (fetched the first time, about 210 MB). The documents are
invented samples that `graphme.samples` writes.

```python
import os
from pathlib import Path

import graphme
from graphme.samples import write_samples

# Your provider and its key. Pages and text are sent to it; nothing else leaves this machine.
SETTINGS = {"provider": "openai", "api_key": os.environ["OPENAI_API_KEY"]}
```

```python
folder = Path("samples")
write_samples(folder)

g = graphme.GraphMe(workspace="Samples", **SETTINGS)
report = g.add(folder)
g.embed()
report
```

```python
answer = g.ask("What is the seat leak test pressure?")
print(answer.text)
for source in answer.sources:
    print(source.file, source.score)
```

```python
for source in g.search("Which work order did the leak failures come from?"):
    print(source.file, [link.file for link in source.linked])
```

```python
g.close()
```
